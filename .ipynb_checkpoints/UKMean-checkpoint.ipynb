```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score

class UKMeans:
    def __init__(self, n_clusters=3, max_iter=100, tol=1e-4, random_state=None):
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tol = tol
        self.random_state = random_state
        np.random.seed(random_state)

    def expected_sq_distance_line(self, centroid, segment_start, segment_end):
        """
        Tính E[||c - x||^2] cho uncertainty dạng đoạn thẳng với phân phối đồng mức.
        Áp dụng công thức: E(||c - x||^2) = C + (B/2) + (D^2/3)
        với:
            D^2 = (c - a)^2 + (d - b)^2,
            B = 2 * ((c - a)*(a - p) + (d - b)*(b - q)),
            C = (p - a)^2 + (q - b)^2,
        với centroid c = (p,q) và đoạn uncertainty có đầu mút (a, b) và (c, d)
        """
        a, b = segment_start
        c_seg, d = segment_end
        p, q = centroid
        
        D_sq = (c_seg - a)**2 + (d - b)**2
        B = 2 * ((c_seg - a) * (a - p) + (d - b) * (b - q))
        C_val = (p - a)**2 + (q - b)**2
        
        return (D_sq / 3) + (B / 2) + C_val

    def fit(self, X, segments):
        # Khởi tạo centroid sử dụng K-means trên dữ liệu recorded
        # (sử dụng khởi tạo k-means++ với n_init=1 để có kết quả nhất định theo random_state)
        kmeans_init = KMeans(n_clusters=self.n_clusters, init='k-means++', n_init=1, random_state=self.random_state)
        kmeans_init.fit(X)
        centroids = kmeans_init.cluster_centers_
        
        for itr in range(self.max_iter):
            # Tính khoảng cách kỳ vọng: với mỗi điểm và mỗi centroid
            distances = np.zeros((X.shape[0], self.n_clusters))
            for j in range(self.n_clusters):
                # Với mỗi điểm, tính khoảng cách kỳ vọng giữa centroid j và uncertainty của điểm đó
                distances[:, j] = [self.expected_sq_distance_line(centroids[j], seg[0], seg[1]) for seg in segments]
            
            labels = np.argmin(distances, axis=1)
            
            # Cập nhật centroid: trung bình các giá trị kỳ vọng (tính là trung điểm của đoạn uncertainty)
            new_centroids = []
            for j in range(self.n_clusters):
                # Lấy các đoạn uncertainty của các điểm được gán vào cluster j
                cluster_segs = [seg for seg, lab in zip(segments, labels) if lab == j]
                if len(cluster_segs) > 0:
                    mid_points = np.array([[(seg[0][0] + seg[1][0]) / 2.0, (seg[0][1] + seg[1][1]) / 2.0] for seg in cluster_segs])
                    new_centroid = np.mean(mid_points, axis=0)
                else:
                    new_centroid = centroids[j]
                new_centroids.append(new_centroid)
            new_centroids = np.array(new_centroids)
            
            # Kiểm tra điều kiện hội tụ dựa trên norm của sự thay đổi centroid
            if np.linalg.norm(new_centroids - centroids) < self.tol:
                centroids = new_centroids
                break
            centroids = new_centroids
        
        self.cluster_centers_ = centroids
        self.labels_ = labels
        return self

def generate_data(n=1000, d=5, random_state=None):
    """
    Tạo dữ liệu mô phỏng với uncertainty dạng đoạn thẳng.
    - recorded: tọa độ ghi nhận (trong không gian 100x100)
    - segments: mỗi điểm có uncertainty được biểu diễn dưới dạng một đoạn thẳng với độ dài chính xác d và hướng ngẫu nhiên
    - actual: vị trí thực tế được lấy ngẫu nhiên trên đoạn uncertainty
    """
    np.random.seed(random_state)
    recorded = np.random.rand(n, 2) * 100  # Dữ liệu ghi nhận
    
    # Tạo uncertainty dạng đoạn thẳng với độ dài d và hướng ngẫu nhiên
    angles = np.random.rand(n) * 2 * np.pi
    segments = []
    actual = []
    for i in range(n):
        dx = d * np.cos(angles[i])
        dy = d * np.sin(angles[i])
        segment_end = recorded[i] + np.array([dx, dy])
        segments.append((recorded[i], segment_end))
        
        # Vị trí thực tế được chọn ngẫu nhiên trên đoạn thẳng
        t = np.random.uniform()
        actual_point = recorded[i] + t * (segment_end - recorded[i])
        actual.append(actual_point)
    
    return recorded, np.array(segments), np.array(actual)

def run_experiment(n=1000, K=20, d=5, n_runs=50, random_state=42):
    """
    Chạy nhiều lần thử nghiệm (n_runs) với các phương pháp:
      (1) K-means trên dữ liệu recorded
      (2) UK-means sử dụng recorded + uncertainty
      (3) K-means trên dữ liệu actual (ground truth)
      
    Sau đó so sánh ARI giữa UK-means và K-means (trên recorded) so với kết quả trên dữ liệu actual.
    """
    ari_uk = []
    ari_kmeans = []
    
    for run in range(n_runs):
        seed = random_state + run
        recorded, segments, actual = generate_data(n=n, d=d, random_state=seed)
        
        # Đảm bảo cùng initial centroids cho cả 3 phương pháp
        kmeans_init = KMeans(n_clusters=K, init='k-means++', n_init=1, random_state=seed)
        kmeans_init.fit(recorded)
        init_centroids = kmeans_init.cluster_centers_
        
        # (1) K-means trên dữ liệu recorded
        kmeans_recorded = KMeans(n_clusters=K, init=init_centroids, n_init=1, random_state=seed).fit(recorded)
        
        # (2) UK-means
        ukmeans = UKMeans(n_clusters=K, max_iter=100, tol=1e-4, random_state=seed).fit(recorded, segments)
        
        # (3) K-means trên dữ liệu actual (ground truth)
        kmeans_actual = KMeans(n_clusters=K, init=init_centroids, n_init=1, random_state=seed).fit(actual)
        
        # Tính ARI giữa phân lớp của UK-means (và K-means trên recorded) với phân lớp tham chiếu (actual)
        ari_uk.append(adjusted_rand_score(kmeans_actual.labels_, ukmeans.labels_))
        ari_kmeans.append(adjusted_rand_score(kmeans_actual.labels_, kmeans_recorded.labels_))
    
    return np.mean(ari_uk), np.mean(ari_kmeans)

# Chạy thử nghiệm với các giá trị d khác nhau
d_values = [1.5, 2.5, 5, 7.5, 10, 20, 50]
results = {}

print("Kết quả ARI (trung bình theo {} lần chạy):".format(50))
for d in d_values:
    uk_ari, km_ari = run_experiment(d=d, n_runs=50)  # Bạn có thể tăng n_runs thành 500 để tái hiện chính xác hơn kết quả trong bài báo
    results[d] = (uk_ari, km_ari)
    print(f"d={d}: UK-means ARI={uk_ari:.3f}, K-means ARI={km_ari:.3f}")

# In bảng kết quả cải tiến
print("\nBảng kết quả cải tiến:")
print("D\tUK-means\tK-means\t\t% Cải thiện")
for d in d_values:
    uk, km = results[d]
    improvement = ((uk - km) / km) * 100 if km != 0 else 0
    print(f"{d}\t{uk:.3f}\t\t{km:.3f}\t\t{improvement:.1f}%")
```
